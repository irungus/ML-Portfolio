# 1.Variables and Data Types
A variable is a named storage location that holds a value or data 
 - Text Type: str
 - Numeric Types: int, float, complex
 - Sequence Types: list, tuple, range
 - Mapping Type: dict
 - Set Types: set, frozenset
 - Boolean Type: bool
 - Binary Types: bytes, bytearry, memoryview
 - None Type: NoneType


# Python Data Types

Python has several built-in data types used to store different kinds of data.

## i. Text Type

- `str` — String
  - Represents text or a sequence of characters.
  - Example: `"Hello World"`

## ii. Numeric Types

- `int` — Integer
  - Represents whole numbers, positive or negative.
  - Example: `25`, `-10`, `0`

- `float` — Floating-point number
  - Represents numbers containing decimal points.
  - Example: `25.5`, `3.14`, `-2.5`

- `complex` — Complex number
  - Represents numbers with a real and imaginary part.
  - Example: `3 + 4j`

## iii. Sequence Types

- `list` — List
  - Represents an ordered and changeable collection of items.
  - Allows duplicate values.
  - Example: `[1, 2, 3]`

- `tuple` — Tuple
  - Represents an ordered and unchangeable collection of items.
  - Allows duplicate values.
  - Example: `(1, 2, 3)`

- `range` — Range
  - Represents a sequence of numbers.
  - Commonly used with loops.
  - Example: `range(5)`

## iv. Mapping Type

- `dict` — Dictionary
  - Represents data as key-value pairs.
  - Example: `{"name": "Samuel", "age": 30}`

## v. Set Types

- `set` — Set
  - Represents an unordered collection of unique values.
  - Duplicate values are automatically removed.
  - Example: `{1, 2, 3}`

- `frozenset` — Frozen Set
  - Represents an immutable set of unique values.
  - Example: `frozenset({1, 2, 3})`

## vi. Boolean Type

- `bool` — Boolean
  - Represents one of two logical values: `True` or `False`.
  - Example: `True`

## vii. Binary Types

- `bytes` — Bytes
  - Represents an immutable sequence of bytes.
  - Example: `b"Hello"`

- `bytearray` — Byte Array
  - Represents a mutable sequence of bytes.
  - Example: `bytearray(b"Hello")`

- `memoryview` — Memory View
  - Represents a view of data stored in another binary object.
  - Allows access to binary data without copying it.
  - Example: `memoryview(b"Hello")`

## viii. None Type

- `NoneType` — None
  - Represents the absence of a value.
  - The only value of this type is `None`.
  - Example: `None`

# Checking Data Types

You can use the `type()` function to determine the data type of a value.


```python
name = "Samuel Irungu"
val = 20
val2 = 20.5
val3 = 3j
lst = ["apple", "banana", "cherry"]
tpl = ("apple", "banana", "cherry")
rng = range(6)
dct = {"name": "Samuel", "age": 36}
st = {"apple", "banana", "cherry"}
fst = frozenset({"apple", "banana", "cherry"})
bl = True
byt = b"Hello"
bytarr = bytearray(5)
mmview = memoryview(bytes(5))
Nn = None
```

```python
# String
print(name)
# integer, float, complex
print(val,val2,val3, sep="\n")
# list, tuple , range
print(lst, tpl, rng, sep="\n")
# Dictionary
print(dct)
# Set, frozenset
print(st,fst, sep="\n")
#Boolean
print(bl)
# Bytes, bytearray, memoryview
print(byt, bytarr, mmview, sep="\n")
#None
print(Nn)
```

```python
# String
print(f"name is of:{type(name)}")
# integer, float, complex
print(f"val is of: {type (val)}", f"val2 is of: {type(val2)}", f"val3 is of: {type(val3)}", sep="\n")
# List, Tuple, Range
print(f"lst is of: {type(lst)}", f"tpl is of: {type(tpl)}", f"rng is of: {type(rng)}", sep="\n")

# Dictionary
print(f"dct is of: {type(dct)}")

# Set, Frozenset
print(f"st is of: {type(st)}", f"fst is of: {type(fst)}", sep="\n")

# Boolean
print(f"bl is of: {type(bl)}")

# Bytes, Bytearray, Memoryview
print(f"byt is of: {type(byt)}", f"bytarr is of: {type(bytarr)}", f"mmview is of: {type(mmview)}", sep="\n")

# None
print(f"Nn is of: {type(Nn)}")
```
